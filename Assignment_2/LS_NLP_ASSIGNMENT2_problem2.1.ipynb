```python
import math
from collections import Counter

def compute_tf(doc):
    tf_dict = {}
    word_count = len(doc)
    for word in doc:
        tf_dict[word] = tf_dict.get(word, 0) + 1

    for word in tf_dict:
        tf_dict[word] /= word_count
    return tf_dict

def compute_idf(corpus):
    N = len(corpus)
    idf_dict = {}
    all_words = set(word for doc in corpus for word in doc)

    for word in all_words:
        count = sum(1 for doc in corpus if word in doc)
        idf_dict[word] = math.log(N / (count)) + 1
    return idf_dict

def compute_tfidf(corpus):
    tfidf_list = []
    idf = compute_idf(corpus)
    for doc in corpus:
        tf = compute_tf(doc)
        tfidf = {word: tf[word] * idf[word] for word in doc}
        tfidf_list.append(tfidf)
    return tfidf_list

corpus = [
    'the sun is star'.split(),
    'the moon is satellite'.split(),
    'the sun and moon are celestial bodies'.split()
]

tfidf_result = compute_tfidf(corpus)

for i, doc in enumerate(tfidf_result):
    print(f"Doc {i+1}:")
    for word, score in doc.items():
        print(f"  {word}: {score:.4f}")
```
